## Disease Background: 

* Prevalence & incidence

Prevalence: In 2025, 7.2 million Americans age 65 or older were diagnosed with Alzheimer's Disease. Approximately 11% of the population 65 and older.[1]
    
Incidence: In 2011, there were 910,000 new cases among Americans older than 65.[1]

* Economic burden

In 2025, there was an estimated $384 billion total payments in care for adults 65 and older needing care for Alzheimer's and other Dementia related ailments[2]. There are both direct and indirect costs associated with caring for an Alzheimer's patient. Direct costs are in the form of healthcare services, residential care, pharmaceuticals and formal social support for the patient. Indirect costs are costs due to lack of formal care; including, the economic value of unpaid informal care, productivity losses and income foregone by patients or caregivers.[2]

* Risk factors (genetic, lifestyle)

Age, head injury, family history, genetics, and certain medical conditions all increase the risk of having Alzheimer's when older[3]. Older individuals, specifically those over 65, are at an increased risk than ages younger than that. Someone who has a severe injury or multiple head injuries and/or certain chronic conditions have a higher likelihood for developing Alzheimer's. If there is a family history of Alzheimer's or Dementia, or if someone has the APP, PSEN1, PSEN2 or APOE ε4 allele, the chance of developing Alzheimer's is significantly heightened.[3]

* Societal determinants

Education: association between lower level education and poor brain health[4]. Higher education is linked with greater use of the brain. When the brain is used frequently, it can become more difficult for the plaques to form which lessens and/or pushes back the effects of Alzheimer's.
    
Access to Health Care: association between lack of access and worse overall health[4]. Lack of access means people don't get the treatment they need when they immediately need it. Treating and seeking help for ailments when they occur ensures the body remains healthy with no adverse side effects after the disease. Chronic conditions also linked with poorer brain health.[4]
    
Physical environment: developed environments linked with better physical activity and healthier choices overall[4]. Being physically active increases overall health and eating healthy gives the body proper nutrients to stay healthy. Having poor health raises risk of chronic conditions, including Alzheimer's Disease.[4]

Loneliness and Social Isolation: social isolation can increase premature death for all causes and is liked with higher dementia risk[4]. Humans are a social species and social interaction increases brain activity which can increase bodily output. The brain is the control center of the body and when it is healthy (besides with chronic condition or physical disability), the body will function better. [4]

* Symptoms

Symptoms of Alzheimer's disease are memory loss, trouble with thinking and reasoning, difficulty making judgments or decisions, struggle to plan and perform familiar tasking, and changes in personality and behavior[5]. These personality and behavior changes include but are not limited to depression, loss of interest in activities, social withdrawal, mood swings, lack of trust in others, aggression, changes in sleep, wandering, loss of inhibitions and delusions[5]. All symptoms do not start appearing at the same time, it is a gradual change that happens over time.

* Diagnosis

Doctors take a patient history from patient and a family member or friend. They also conduct memory/brain function tests and order blood tests, urine, collect cerebrospinal fluid via a lumbar puncture and perform brain scans. Additionally, they administer a psychiatric evaluation to rule our other mental health conditions that could be causing or contributing to the symptoms. [6]

* Standard of care treatments (& reimbursement) 

Unfortunately, there is no known cure for Alzheimer’s disease, so treatment mainly focuses on treating symptoms and slowing cognitive decline. In the early stages, doctors often prescribe cholinesterase inhibitors like donepezil which prevent the breakdown of acetylcholine (ACh), a compound that helps communication between nerve cells and overall memory. As Alzheimer’s progresses to the moderate to severe stages, medication like memantine is used to regulate glutamate, since too much glutamate can overstimulate and kill nerve cells through a process called excitotoxicity. Both drugs can help for a while but lose effectiveness as the disease progresses [1]. Newer anti-amyloid antibodies, lecanemab and donanemab, were approved for early Alzheimer’s and slowed cognitive decline in some patients over 18 months, but they can cause swelling or bleeding in the brain, known as ARIA [1]. This risk is highest in those with the APOE e4 gene we looked at, so doctors recommended testing before treatment [2]. Medicare covers these drugs if doctors report patient data to a national registry [3], but they are absurdly expensive, starting at roughly 26,500 to 32,000 dollars per year [4].

* Disease progression & prognosis

Alzheimer’s happens slowly and symptomes progress at different rates depending on the case; changes in the brain can appear a decade before symptoms appear [5]. After the onset, the disease moves through three stages: early, middle, and late. Early on, people struggle to find works, misplace things, and get confused with time/place, though they can still work and drive. The middle stage tends to be the longest and brings personality changes, like paranoia or depression, and people start to forger basic facts, like addresses and names. Here, patients need help with daily tasks. In the late stage, people lose the ability to hold a conversation or control movement and need round-the-clock care [6]. Memory also tends to burn backwards, meaning patients lose their newest memories first. Someone may tell clear stories from childhood but not know what day it is or what they ate for breakfast. On average, people live four to eight years after diagnosis, though some live as long as 20 [6]. After death, pathologists measure how far the disease spread using the Braak stage (tau) and Thal phase (amyloid), both of which are in our dataset. 

* Continuum of care providers

Care usually starts with a primary care physician who notices memory problems and gives a quick screening test like the MMSE, the cognitive test recorded in our dataset. Patients are then often referred to neurologists, geriatricians, or neuropsychologists for a full diagnosis. As the disease progresses, the care team expands to nurses, social workers, therapists, and home health aides, and many patients eventually move into memory care or nursing homes, with hospice at the end of life. Most of the care, however, comes from family and friends: nearly 13 million people provided more than 19 billion hours of unpaid care in the past year [7]. For the donors in our dataset, the continuum ended with brain donation to research, which made this project possible.

* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)

Damage starts in the hippocampus and entorhinal cortex, regions involved in memory, which is why memory loss is usually the first symptom. From there the disease spreads thorughout the brain until damage is widespread, shrinking the brain significantly over time. The primary biological features of the disease are the accumaltion specific proteins in the brain, amyloid and tau, working with age-related symptoms like inflammation. Amyloid-Beta forms plaques that disrupt chemical signalling between neurons, while tau tangles destory the internal transport system, leading to neuron death [5]. Our dataset measured both proteins directly in brain tissue as ABeta42 (a form of beta-amyloid) and pTAU (a phosphorylated form of tau). In our data, higher pTAU was significantly correlated with lower MMSE scores (r = -0.29, p = 0.009), and APOE e4 carriers had significantly higher pTAU than non-carriers (p = 0.024).

* Clinical Trials/next-gen therapies

Alzheimer’s treatment is one the largest areas of research and continues to grow. The 2026 drug development pipeline report found 158 drugs in 192 clinical trials, with about 73% aiming to change the course of the disease rather than just treat symptoms. The focus is also moving away from amyloid. Over the past decade, drugs targeting tau and drugs targeting inflammation each grew from about 6% to 20% of the pipeline, while amyloid drugs fell from 33% to about 20% [8]. Trust in amyloid research also took a hit in 2022, when a Science investigation titled "Blots on a Field?" found signs of image manipulation in an influential 2006 amyloid paper that many labs had built their work on [9]. Since pTAU, not ABeta42, was linked to MMSE scores in our data, the growing interest in tau-targeted therapies fits with what we found.

## Data Analysis: 
The goal was to see how Tau pathology relates to genetic risk and cognition in Alzheimer's disease. We asked these two questions:

1. Do APOE e4 carriers have higher pTAU levels in their brains than non-carriers?
2. Do donors with higher pTAU score lower on the MMSE?

To answer them, we wrote a `Patient` class (in `patient_final_Vince.py`) that turns each row of the CSV into a patient object with attributes like sex, age at death, APOE genotype, MMSE score, Braak stage, and pTAU. The class also has methods that group patients for each analysis. We used `statistics` for means and standard deviations, `scipy` for the statistical tests, `scikit-learn` for linear regression, and `matplotlib` for the figures.

* To run this notebook, keep `patient_final_Vince.py` and the CSV file in the same folder as this notebook.

```python
from patient_final_Vince import *
import matplotlib.pyplot as plt
import statistics
from scipy import stats
from sklearn.linear_model import LinearRegression
```

#### 1. Loading Our Patient Data from CSV

Each row of the CSV is converted into a `Patient` object using the `instantiate_from_csv` class method. Text scores like "Braak IV" are converted into integers, and blank cells are stored as `None` so they can be skipped later. We clear the patient list first so that re-running this cell doesn't load the same patients twice.

```python
# Clear the list so re-running this cell doesn't add duplicate patients
Patient.all_patients.clear()

# Create one Patient object per row of the CSV
Patient.instantiate_from_csv("Metadata and Protein Data for Module 1.csv")

# Check that all patients loaded
print(f"Number of patients = {len(Patient.all_patients)}")   # should be 84
print(Patient.all_patients[0])
```

#### 2. Checking Data for Errors

Before graphing, we checked for missing or outlier values. The pTAU measurements were complete for all 84 donors, but 4 donors had no MMSE score, and one donor (H20.33.002) had an MMSE score of 33, which is impossible since the test is scored out of 30. We excluded these 5 donors from the MMSE analysis using the `get_valid_mmse_patients()` method.

For APOE, we counted the number of e4 alleles in each genotype and grouped donors with at least one e4 allele (3_4, 4_4, and 2_4) as carriers.

```python
# Finding patients with a missing or impossible MMSE score
for patient in Patient.all_patients:
    if patient.mmse is None or patient.mmse > Patient.MAX_MMSE:
        print(f"Excluded from MMSE analysis: {patient.donor_id} (MMSE = {patient.mmse})")

print(f"Patients with a usable MMSE score = {len(Patient.get_valid_mmse_patients())}")

# Counting APOE e4 carriers and non-carriers
print(f"APOE e4 carriers = {len(Patient.filter_by_apoe4(carrier=True))}")
print(f"Non-carriers = {len(Patient.filter_by_apoe4(carrier=False))}")
```

#### 3. pTAU in APOE e4 Carriers vs. Non-Carriers - Bar Graph

APOE e4 is the strongest genetic risk factor for Alzheimer's, so we wanted to see if carriers had more tau pathology in their brains. To test this, donors were split into carriers and non-carriers, then the mean and standard deviation of pTAU for each group was calculated.

To test whether the difference was significant, we used Welch's t-test, which does not assume the two groups have equal variance. This mattered here because the groups were different sizes (25 vs. 59 donors). We also ran a one-way ANOVA to match the analysis from Module 1.

- *Null hypothesis:* APOE e4 carriers and non-carriers have the same mean pTAU level.
- *Significance level:* p < 0.05

```python
# --- 3a: Splitting patients into the two groups ---
ptau_e4_carriers = []
ptau_non_carriers = []

for patient in Patient.filter_by_apoe4(carrier=True):
    ptau_e4_carriers.append(patient.ptau)
for patient in Patient.filter_by_apoe4(carrier=False):
    ptau_non_carriers.append(patient.ptau)

# --- 3b: Calculating the mean and standard deviation of each group ---
x_e4_carrier_bar = statistics.mean(ptau_e4_carriers)
x_non_carrier_bar = statistics.mean(ptau_non_carriers)
ptau_e4_carrier_stdev = statistics.stdev(ptau_e4_carriers)
ptau_non_carrier_stdev = statistics.stdev(ptau_non_carriers)

print(f"APOE e4 carriers: mean = {x_e4_carrier_bar:.2f}, stdev = {ptau_e4_carrier_stdev:.2f}")
print(f"Non-carriers: mean = {x_non_carrier_bar:.2f}, stdev = {ptau_non_carrier_stdev:.2f}")

# --- 3c: Testing whether the two group means are different ---
t_stat, p_value = stats.ttest_ind(ptau_e4_carriers, ptau_non_carriers, equal_var=False)
f_statistic, anova_p_value = stats.f_oneway(ptau_e4_carriers, ptau_non_carriers)
print(f"Welch's t-test: t = {t_stat:.2f}, p = {p_value:.3f}")
print(f"One-way ANOVA: F = {f_statistic:.2f}, p = {anova_p_value:.3f}")

# --- 3d: Plotting the bar graph ---
apoe_groups = [f"APOE e4 Carriers (n={len(ptau_e4_carriers)})", f"Non-Carriers (n={len(ptau_non_carriers)})"]
mean_ptau = [x_e4_carrier_bar, x_non_carrier_bar]
stdev_ptau = [ptau_e4_carrier_stdev, ptau_non_carrier_stdev]

plt.figure()   # start a new, blank graph
plt.bar(apoe_groups, mean_ptau, yerr=stdev_ptau, capsize=10, color=["purple", "gray"])

# Placing the test results just above the taller error bar
top_of_bars = max(x_e4_carrier_bar + ptau_e4_carrier_stdev,
                  x_non_carrier_bar + ptau_non_carrier_stdev)
plt.text(0.5, top_of_bars * 1.05,
         f"Welch's t-test: t = {t_stat:.2f}, p = {p_value:.3f}\n"
         f"ANOVA: F = {f_statistic:.2f}, p = {anova_p_value:.3f}", ha="center")
plt.ylim(0, top_of_bars * 1.25)   # extra room so the text isn't cut off

plt.title("Average pTAU Levels in APOE e4 Carriers vs. Non-Carriers")
plt.xlabel("APOE Group")
plt.ylabel("Average pTAU (pg/ug)")
plt.show()
```

#### 4. pTAU vs. MMSE Score - Scatterplot

For our second graph, we wanted to see how levels of pTAU in the brain relates to cognitive ability. For every donor with a usable MMSE score, we paired their pTAU level with their last MMSE score. Then, we fit a linear regression model with  `scikit-learn` to get the slope, intercept, and R-squared. Since `scikit-learn` doesn't give a p-value, we also calculated the Pearson correlation coefficient (r) and its p-value with `scipy`.

- *Null hypothesis:* There is no linear relationship between pTAU and MMSE score.
- *Significance level:* p < 0.05

```python
# --- 4a: Collecting paired pTAU and MMSE values ---
# pTAU and MMSE are added in the same step so the pairs stay lined up
ptau_levels = []
mmse_scores = []

for patient in Patient.get_valid_mmse_patients():
    ptau_levels.append(patient.ptau)
    mmse_scores.append(patient.mmse)

x = ptau_levels   # Independent variable: pTAU
y = mmse_scores   # Dependent variable: MMSE score

# --- 4b: Fitting a linear regression line ---
X = [[value] for value in x]   # scikit-learn expects a two-dimensional list
model = LinearRegression()
model.fit(X, y)
slope = model.coef_[0]
intercept = model.intercept_
regression_r_squared = model.score(X, y)
regression_x = [min(x), max(x)]
regression_y = model.predict([[value] for value in regression_x])

print(f"Linear regression: slope = {slope:.4f}, intercept = {intercept:.2f}, R-squared = {regression_r_squared:.3f}")

# --- 4c: Testing whether the correlation is significant ---
r_value, r_p_value = stats.pearsonr(x, y)
print(f"Pearson correlation: r = {r_value:.2f}, p = {r_p_value:.3f}, n = {len(x)}")

# --- 4d: Plotting the scatter plot ---
plt.figure()   # start a new, blank graph
plt.scatter(x, y, color="purple")
plt.plot(regression_x, regression_y, color="black", linewidth=2, label="Linear regression")
plt.legend(loc="lower right")

# Printing the results in the top right corner of the graph
plt.text(0.97, 0.97,
         f"r = {r_value:.2f}, p = {r_p_value:.3f}\n"
         f"R-squared = {regression_r_squared:.3f}\n"
         f"n = {len(x)}",
         transform=plt.gca().transAxes, ha="right", va="top")

plt.title("pTAU Levels vs. MMSE Score")
plt.xlabel("pTAU (pg/ug)")
plt.ylabel("MMSE Score")
plt.show()
```

## Conclusions and Ethical Implications: 

Our project looked at how tau pathology relates to cognitive decline and genetic risk factors in Alzheimer’s disease. First, we found that APOE e4 carriers displayed higher pTAU than non-carriers (5.24 vs. 3.77 pg/ug; Welch’s t-test p = 0.024, ANOVA p = 0.016), which supports the established evidence that APOE e4 is the strongest genetic risk factor for Alzheimer’s. Second, donors with more pTAU tended to have lower MMSE scores (r = -0.29, p = 0.009). However, the R-squared value of 0.086 means pTAU explained less than 10% of the variation in MMSE scores, so tau appears to be a factor, but it is far from the whole story. The rest of our data showed something similar: 8 highly educated donors with severe tau pathology did not have dementia. Overall, our results support the growing focus on tau as a target for treatment while also showing that brain pathology alone cannot predict how well cognition will hold up.

It’s important not to overlook the ethical implications of a study like this, especially when it comes to Alzheimer’s, a disease that has affected so many people. Since the APOE e4 gene is tied to both disease risk and treatment safety, more people may seek genetic testing, which can cause anxiety and raise privacy concerns. In the U.S., the Genetic Information Nondiscrimination Act (GINA) protects against genetic discrimination in health insurance and employment, but not in life, disability, or long-term care insurance. It is also important to remember that being an APOE e4 carrier or having high tau does not guarantee that someone will develop dementia. These factors increase risk, but no single factor that we know of directly causes dementia. Making claims that suggest otherwise could lead to unnecessary fear. Lastly, 81 of our 84 donors were White, and the average donor had over 16 years of education. Our conclusions may not apply to groups that are often more affected by Alzheimer’s but are underrepresented in research, and using data like this to guide care could potentially widen existing health disparities.


## Limitations and Future Work: 

Our analysis had several limitations. The sample was relatively small, with only 84 donors and 25 APOE e4 carriers, and the pTAU vs. MMSE correlation was influenced in part by a few donors with very high pTAU. When those outliers were removed, the relationship became weaker. Timing was also an issue, since pTAU was measured in brain tissue after death, while the last MMSE was taken a median of about 28 months earlier, and in some cases almost 10 years earlier. A patient’s cognition almost certainly changed during that gap, making it harder to connect their final brain pathology to their recorded cognitive score. We also looked at several different variable pairs before deciding which ones to graph, which increases the chance that a significant result occurred by chance. Lastly, we did not control for other factors such as age, sex, education, or co-pathologies. Because this was a correlational analysis, we also cannot conclude that tau directly causes cognitive decline.

Future work could address these limitations by using a larger and more diverse dataset and by only including donors whose MMSE was taken close to death. A multiple regression model could test whether pTAU still predicts MMSE after controlling for age, sex, education, and APOE status. Splitting APOE status into 0, 1, or 2 copies of the e4 allele could also show whether risk changes with each additional copy. Finally, it would be interesting to repeat this analysis using the blood-based pTau217 tests now used in the clinic. This could help connect post-mortem findings like ours to measurements that can be taken from living patients.

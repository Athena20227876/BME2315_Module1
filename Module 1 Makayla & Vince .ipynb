## Disease Background: 

* Prevalence & incidence

Prevalence: In 2025, 7.2 million Americans age 65 or older were diagnosed with Alzheimer's Disease. Approximately 11% of the population 65 and older.
    
Incidence: In 2011, there were 910,000 new cases among Americans older than 65.

Link: pmc.ncbi.nlm.nih.gov/articles/PMC12040760/#alz70235-sec-0530

* Economic burden

In 2025, there was an estimated $384 billion total payments in care for adults 65 and older needing care for Alzheimer's and other Dementia related ailments. There are both direct and indirect costs associated with caring for an Alzheimer's patient. Direct costs are in the form of healthcare services, residential care, pharmaceuticals and formal social support for the patient. Indirect costs are costs due to lack of formal care; including, the economic value of unpaid informal care, productivity losses and income foregone by patients or caregivers.

Link: www.nature.com/nature-index/topics/l4/economic-burden-of-alzheimer-s-disease-and-dementia-care 

* Risk factors (genetic, lifestyle)

Age, head injury, family history, genetics, and certain medical conditions all increase the risk of having Alzheimer's when older.

Link: www.alz.org/alzheimers-dementia/what-is-alzheimers/causes-and-risk-factors

Education: association between lower level education and poor brain health.

Loneliness and Social Isolation: social isolation can increase premature death for all causes and is liked with higher dementia risk.

* Symptoms

Symptoms of Alzheimer's disease are memory loss, trouble with thinking and reasoning, difficulty making judgments or decisions, struggle to plan and perform familiar tasking, and changes in personality and behavior. These personality and behavior changes include but are not limited to depression, loss of interest in activities, social withdrawal, mood swings, lack of trust in others, aggression, changes in sleep, wandering, loss of inhibitions and delusions.

Link: www.mayoclinic.org/diseases-conditions/alzheimers-disease/symptoms-causes/syc-20350447?os=xd4wd&p=1

* Diagnosis

Doctors take a patient history from patient and a family member or friend. They also conduct memory/brain function tests and order blood tests, urine, collect cerebrospinal fluid via a lumbar puncture and perform brain scans. Additionally, they administer a psychiatric evaluation to rule our other mental health conditions that could be causing or contributing to the symptoms.

Link: www.nia.nih.gov/health/alzheimers-symptoms-and-diagnosis/how-alzheimers-disease-diagnosed

* Standard of care treatments (& reimbursement) 

Unfortunately, there is no known cure for Alzheimer’s disease, so treatment mainly focuses on treating symptoms and slowing cognitive decline. In the early stages, doctors often prescribe cholinesterase inhibitors like donepezil which prevent the breakdown of acetylcholine (ACh), a compound that helps communication between nerve cells and overall memory. As Alzheimer’s progresses to the moderate to severe stages, medication like memantine is used to regulate glutamate, since too much glutamate can overstimulate and kill nerve cells through a process called excitotoxicity. Both drugs can help for a while but lose effectiveness as the disease progresses [1]. Newer anti-amyloid antibodies, lecanemab and donanemab, were approved for early Alzheimer’s and slowed cognitive decline in some patients over 18 months, but they can cause swelling or bleeding in the brain, known as ARIA [1]. This risk is highest in those with the APOE e4 gene we looked at, so doctors recommended testing before treatment [2]. Medicare covers these drugs if doctors report patient data to a national registry [3], but they are absurdly expensive, starting at roughly 26,500 to 32,000 dollars per year [4].

* Disease progression & prognosis

Alzheimer’s happens slowly and symptomes progress at different rates depending on the case; changes in the brain can appear a decade before symptoms appear [5]. After the onset, the disease moves through three stages: early, middle, and late. Early on, people struggle to find works, misplace things, and get confused with time/place, though they can still work and drive. The middle stage tends to be the longest and brings personality changes, like paranoia or depression, and people start to forger basic facts, like addresses and names. Here, patients need help with daily tasks. In the late stage, people lose the ability to hold a conversation or control movement and need round-the-clock care [6]. Memory also tends to burn backwards, meaning patients lose their newest memories first. Someone may tell clear stories from childhood but not know what day it is or what they ate for breakfast. On average, people live four to eight years after diagnosis, though some live as long as 20 [6]. After death, pathologists measure how far the disease spread using the Braak stage (tau) and Thal phase (amyloid), both of which are in our dataset. 

* Continuum of care providers

Care usually starts with a primary care physician who notices memory problems and gives a quick screening test like the MMSE, the cognitive test recorded in our dataset. Patients are then often referred to neurologists, geriatricians, or neuropsychologists for a full diagnosis. As the disease progresses, the care team expands to nurses, social workers, therapists, and home health aides, and many patients eventually move into memory care or nursing homes, with hospice at the end of life. Most of the care, however, comes from family and friends: nearly 13 million people provided more than 19 billion hours of unpaid care in the past year [7]. For the donors in our dataset, the continuum ended with brain donation to research, which made this project possible.

* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)

Damage starts in the hippocampus and entorhinal cortex, regions involved in memory, which is why memory loss is usually the first symptom. From there the disease spreads thorughout the brain until damage is widespread, shrinking the brain significantly over time. The primary biological features of the disease are the accumaltion specific proteins in the brain, amyloid and tau, working with age-related symptoms like inflammation. Amyloid-Beta forms plaques that disrupt chemical signalling between neurons, while tau tangles destory the internal transport system, leading to neuron death [5]. Our dataset measured both proteins directly in brain tissue as ABeta42 (a form of beta-amyloid) and pTAU (a phosphorylated form of tau). In our data, higher pTAU was significantly correlated with lower MMSE scores (r = -0.29, p = 0.009), and APOE e4 carriers had significantly higher pTAU than non-carriers (p = 0.024).

* Clinical Trials/next-gen therapies

Alzheimer’s treatment is one the largest areas of research and continues to grow. The 2026 drug development pipeline report found 158 drugs in 192 clinical trials, with about 73% aiming to change the course of the disease rather than just treat symptoms. The focus is also moving away from amyloid. Over the past decade, drugs targeting tau and drugs targeting inflammation each grew from about 6% to 20% of the pipeline, while amyloid drugs fell from 33% to about 20% [8]. Trust in amyloid research also took a hit in 2022, when a Science investigation titled "Blots on a Field?" found signs of image manipulation in an influential 2006 amyloid paper that many labs had built their work on [9]. Since pTAU, not ABeta42, was linked to MMSE scores in our data, the growing interest in tau-targeted therapies fits with what we found.

## Conclusions and Ethical Implications: 
*(Think about the answer your analysis generated, draw conclusions related to your overarching question, and discuss the ethical implications of your conclusions.*

Our project looked at how tau pathology relates to cognitive decline and genetic risk factors in Alzheimer’s disease. First, we found that APOE e4 carriers displayed higher pTAU than non-carriers (5.24 vs. 3.77 pg/ug; Welch’s t-test p = 0.024, ANOVA p = 0.016), which supports the established evidence that APOE e4 is the strongest genetic risk factor for Alzheimer’s. Second, donors with more pTAU tended to have lower MMSE scores (r = -0.29, p = 0.009). However, the R-squared value of 0.086 means pTAU explained less than 10% of the variation in MMSE scores, so tau appears to be a factor, but it is far from the whole story. The rest of our data showed something similar: 8 highly educated donors with severe tau pathology did not have dementia. Overall, our results support the growing focus on tau as a target for treatment while also showing that brain pathology alone cannot predict how well cognition will hold up.

It’s important not to overlook the ethical implications of a study like this, especially when it comes to Alzheimer’s, a disease that has affected so many people. Since the APOE e4 gene is tied to both disease risk and treatment safety, more people may seek genetic testing, which can cause anxiety and raise privacy concerns. In the U.S., the Genetic Information Nondiscrimination Act (GINA) protects against genetic discrimination in health insurance and employment, but not in life, disability, or long-term care insurance. It is also important to remember that being an APOE e4 carrier or having high tau does not guarantee that someone will develop dementia. These factors increase risk, but no single factor that we know of directly causes dementia. Making claims that suggest otherwise could lead to unnecessary fear. Lastly, 81 of our 84 donors were White, and the average donor had over 16 years of education. Our conclusions may not apply to groups that are often more affected by Alzheimer’s but are underrepresented in research, and using data like this to guide care could potentially widen existing health disparities.


## Limitations and Future Work: 
*(Think about the answer your analysis generated, draw conclusions related to your overarching question, and discuss the ethical implications of your conclusions.*

Our analysis had several limitations. The sample was relatively small, with only 84 donors and 25 APOE e4 carriers, and the pTAU vs. MMSE correlation was influenced in part by a few donors with very high pTAU. When those outliers were removed, the relationship became weaker. Timing was also an issue, since pTAU was measured in brain tissue after death, while the last MMSE was taken a median of about 28 months earlier, and in some cases almost 10 years earlier. A patient’s cognition almost certainly changed during that gap, making it harder to connect their final brain pathology to their recorded cognitive score. We also looked at several different variable pairs before deciding which ones to graph, which increases the chance that a significant result occurred by chance. Lastly, we did not control for other factors such as age, sex, education, or co-pathologies. Because this was a correlational analysis, we also cannot conclude that tau directly causes cognitive decline.

Future work could address these limitations by using a larger and more diverse dataset and by only including donors whose MMSE was taken close to death. A multiple regression model could test whether pTAU still predicts MMSE after controlling for age, sex, education, and APOE status. Splitting APOE status into 0, 1, or 2 copies of the e4 allele could also show whether risk changes with each additional copy. Finally, it would be interesting to repeat this analysis using the blood-based pTau217 tests now used in the clinic. This could help connect post-mortem findings like ours to measurements that can be taken from living patients.
